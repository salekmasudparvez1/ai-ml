# 🔥 Core Patterns (90% of LeetCode Pandas)

Master these patterns and you can solve most problems.

```python
import pandas as pd

# Sample Data
df = pd.DataFrame({
    'employee': ['A', 'B', 'C', 'D', 'E'],
    'department': ['HR', 'HR', 'IT', 'IT', 'IT'],
    'salary': [5000, 6000, 7000, 8000, 8000],
    'age': [25, 30, 35, 40, 28]
})

df
```

## 1️⃣ Filtering Pattern

👉 Used when condition-based selection is needed

Example:
Find employees with salary > 6000


```python
df[df['salary'] > 6000]
```

## 2️⃣ GroupBy + Aggregation

👉 Used when summarizing data

Example:
Average salary per department


```python
df.groupby('department')['salary'].mean()
```

## 3️⃣ Top N / Sorting

👉 Used for ranking problems

Example:
Top 2 highest salaries


```python
df.sort_values('salary', ascending=False).head(2)
```

## 4️⃣ Ranking / Window Functions

👉 Important for interview problems

Example:
Rank employees by salary


```python
df['rank'] = df['salary'].rank(method='dense', ascending=False)
df
```

## 5️⃣ Transform (VERY IMPORTANT)

👉 Used when you need group results back to original rows

Example:
Find highest salary per department AND show employees


```python
max_salary = df.groupby('department')['salary'].transform('max')
df[df['salary'] == max_salary]
```

# 🧪 Real LeetCode Style Problems

## Problem 1: Second Highest Salary

### Thinking Process:
- Remove duplicates
- Sort descending
- Pick second value


```python
df['salary'].drop_duplicates().sort_values(ascending=False).iloc[1]
```

## Problem 2: Duplicate Emails (Concept)

### Thinking:
- Find duplicates
- Count occurrences


```python
df['employee'].value_counts()[lambda x: x > 1]
```
