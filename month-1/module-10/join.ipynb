## 🎯 Goal

Find employees who earn more than their manager


```python
import pandas as pd

df = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["Joe", "Henry", "Sam", "Max"],
    "salary": [70000, 80000, 60000, 90000],
    "managerId": [3, 4, None, None]
})

df
```

# 🔍 Step 1: Understand the Problem

We need:
- Employee salary
- Manager salary

❌ Problem: Manager salary is NOT directly in the row

👉 So we must connect rows together


# 🔗 Step 2: What is merge()?

merge = JOIN (like SQL)

It combines two tables based on matching columns.

### Simple Example


```python
left = pd.DataFrame({
    "id": [1, 2],
    "name": ["A", "B"]
})

right = pd.DataFrame({
    "id": [1, 2],
    "salary": [100, 200]
})

left.merge(right, on="id")
```

👉 This joins tables where `id` matches


# 🔁 Step 3: Self Join Concept

We are merging the SAME table with itself.

Why?
- Employee row has managerId
- Manager is ALSO in same table

👉 So we match:
employee.managerId = manager.id


```python
merged = df.merge(
    df,
    left_on="managerId",   # employee side
    right_on="id",        # manager side
    suffixes=("_emp", "_mgr")
)

merged
```

# 🧾 What happened after merge?

Now we have:

- employee data (suffix _emp)
- manager data (suffix _mgr)

Example row:

| name_emp | salary_emp | name_mgr | salary_mgr |
|----------|-----------|----------|------------|
| Joe      | 70000     | Sam      | 60000      |

🔥 Now we can compare easily!


# ⚖️ Step 4: Filtering Condition

We only want:

employee.salary > manager.salary


```python
result = merged[merged["salary_emp"] > merged["salary_mgr"]]
result
```

# 🧾 Step 5: Select Final Column

We only need employee name


```python
final = result[["name_emp"]].rename(columns={"name_emp": "Employee"})
final
```

# 🔥 Final Answer

Joe earns more than his manager.


# 🧠 Deep Understanding Summary

### Why merge?
Because data you need is in another row

### Why suffixes?
To avoid column name conflict

### Why filtering?
To apply condition after joining
