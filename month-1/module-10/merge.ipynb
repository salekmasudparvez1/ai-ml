# 📘 Pandas DataFrame `merge()` Method
## Beginner to Advanced Guide

Learn how to combine DataFrames using logic similar to SQL JOIN operations. The `merge()` function is the standard for high-performance database-style joining in Python.

```python
import pandas as pd
```

## 1️⃣ Create Sample DataFrames
We define two datasets. Note that **Sally** exists in both, but her **age** is recorded differently. This is a crucial detail for merging logic.

```python
df1 = pd.DataFrame({
    'name': ['Sally', 'Mary', 'John'],
    'age': [50, 40, 30]
})

df2 = pd.DataFrame({
    'name': ['Sally', 'Peter', 'Micky'],
    'age': [77, 44, 22]
})

print("DataFrame 1 (Left):\n", df1)
print("\nDataFrame 2 (Right):\n", df2)
```

## 2️⃣ Basic Right Merge Example
A **Right Merge** returns all rows from the right DataFrame and the matched rows from the left. 

> **Crucial Fix:** We specify `on='name'`. Without this, Pandas tries to match on both 'name' AND 'age', which results in an empty merge for 'Sally' because her ages (50 vs 77) do not match.

```python
right_merged = df1.merge(df2, how='right', on='name')
print(right_merged)
```

## 3️⃣ Types of Merge
Understanding the direction of your join is vital for data integrity.



* **Inner:** Only rows where the key exists in BOTH DataFrames.
* **Left:** All rows from the left, matched rows from the right.
* **Outer:** All rows from both DataFrames (fills missing with `NaN`).

```python
print("INNER JOIN:\n", df1.merge(df2, on='name', how='inner'))
print("\nLEFT JOIN:\n", df1.merge(df2, on='name', how='left'))
print("\nOUTER JOIN:\n", df1.merge(df2, on='name', how='outer'))
```

## 4️⃣ Merge Using a Specific Key Column
Standardizing the join on a unique identifier (like an ID) prevents duplicate row creation.

```python
customers = pd.DataFrame({'id': [1,2,3], 'name': ['Alice','Bob','Charlie']})
orders = pd.DataFrame({'id': [1,2,1], 'product': ['Laptop','Phone','Mouse']})

print(customers.merge(orders, on='id'))
```

## 5️⃣ left_on and right_on Example
Use these when the column names differ between datasets but represent the same data.



```python
a = pd.DataFrame({'cust_id':[1,2], 'name':['A','B']})
b = pd.DataFrame({'client_id':[1,2], 'city':['NY','LA']})

print(a.merge(b, left_on='cust_id', right_on='client_id'))
```

## 6️⃣ Using Suffixes
When non-merged columns have the same name, `suffixes` helps identify which DataFrame they originated from.

```python
print(df1.merge(df2, on='name', suffixes=('_original','_updated')))
```

## 7️⃣ The Indicator Parameter
Adding `indicator=True` creates a column named `_merge` that tells you if a row was found in both, or only one of the DataFrames.

```python
print(df1.merge(df2, on='name', how='outer', indicator=True))
```

## 8️⃣ Index-Based Merge
Sometimes data is aligned by index rather than a column. Use `left_index` and `right_index` for this.

```python
x = pd.DataFrame({'A':[1,2]}, index=['a','b'])
y = pd.DataFrame({'B':[3,4]}, index=['a','b'])

print(x.merge(y, left_index=True, right_index=True))
```

## 9️⃣ Real World Example: Employee Salary Merge
A typical task is combining profile data with financial data.

```python
emp = pd.DataFrame({'emp_id':[1,2,3], 'name':['Tom','Jerry','Mike']})
salary = pd.DataFrame({'emp_id':[1,2,3], 'salary':[5000,6000,7000]})

print(emp.merge(salary, on='emp_id'))
```

## 🔟 Common Mistakes
* **Implicit Merges:** Forgetting `on=` and letting Pandas guess, which often leads to empty DataFrames or unwanted multi-column joins.
* **Type Mismatch:** Trying to merge a string ID column with an integer ID column.
* **Memory Bloat:** Performing a many-to-many join that creates a massive Cartesian product.

## 🎯 Interview Questions
1. **Difference between `merge()` and `join()`?** 
   * `merge()` is column-based and highly flexible; `join()` is index-based by default.
2. **What is an outer merge?** 
   * A union of both DataFrames that keeps all keys from both.
3. **How to merge on different column names?** 
   * Use the `left_on` and `right_on` parameters.
4. **How to detect unmatched rows?** 
   * Use `how='outer'` with `indicator=True`.
