```python
import pandas as pd

customers = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["Joe", "Henry", "Sam", "Max"]
})

orders = pd.DataFrame({
    "id": [1, 2],
    "customerId": [3, 1]
})

print(customers)
print(orders)
```

```python
merged = pd.merge(customers,orders,left_on="id",right_on="customerId",how="inner")
result= merged[["id_x","name"]].rename(columns={
    "id_x":"id",
})
find = customers[["name"]].isin([result["name"]])
find
```
