# 0) Starter Code (Run First)

```python
# You must run this code and use this dataset for all questions
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
num_students = 50

# Student data
student_ids = np.arange(201, 201 + num_students)
student_ages = np.random.randint(15, 22, num_students)
student_scores = np.random.randint(40, 100, num_students)
student_attendance = np.random.randint(60, 100, num_students)  # percentage

departments_list = [
    "Science",
    "Commerce",
    "Arts",
    "Computer Science",
    "Mathematics",
    "English",
    "Physics",
    "Chemistry",
    "Biology",
    "Economics"
]

student_departments = np.random.choice(departments_list, num_students)

grade_conditions = [
    student_scores >= 80,
    (student_scores >= 65) & (student_scores < 80),
    (student_scores >= 50) & (student_scores < 65),
    student_scores < 50
]
grade_values = ["A", "B", "C", "F"]
student_grades = np.select(grade_conditions, grade_values, default="F")

# Create DataFrame
data = pd.DataFrame({
    "student_id": student_ids,
    "student_age": student_ages,
    "score": student_scores,
    "attendance": student_attendance,
    "department": student_departments,
    "grade": student_grades
})

# Save dataset
data.to_csv("school_data.csv", index=False)
```

# Question 02: [ Marks 10 ]

## Using NumPy functions on :

* Find the maximum and minimum score
* Calculate the average score
* Calculate the average attendance

Then, extract students who:

* Have a score greater than the average score

Print:

* The filtered score array

## Write Answer 02:


```python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/school_data.csv")

score_array = df["score"].to_numpy()
attendance_array = df["attendance"].to_numpy()

maximum_score = score_array.max()
minimum_score = score_array.min()
average_score = score_array.mean()
average_attendance = attendance_array.mean()

print("maximum_score =",maximum_score)
print("minimum_score =",minimum_score)
print("average_score =",average_score)
print("average_attendance =",average_attendance)

# print(score_array > average_score)
greater_than_avg_array = score_array[score_array > average_score]
print("greater_than_avg_array=",greater_than_avg_array)
```


# Question 04: [ Marks 10 ]

##
Using .loc:

* Increase the score by 10% for students in "Computer Science" and "Mathematics" departments
* Increase the score by 5% for all other departments
* Cap the score at 100 (no score should exceed 100 after the raise)

Then print a DataFrame that includes:

* student_id, department, previous_score, new_score, score_increase


## Write Answer 04:

```python

new_df = df.copy()

filter_dep = new_df['department'].isin(["Computer Science","Mathematics"])
#type_error_fix
df["score"]=df["score"].astype(float)
new_df["score"]=new_df["score"].astype(float)

filter_rest_dep = ~filter_dep
#increase10%

new_df.loc[filter_dep,'score'] = new_df.loc[filter_dep,'score']+ new_df.loc[filter_dep,'score']*0.1
#increase5%
new_df.loc[filter_rest_dep,'score'] = new_df.loc[filter_rest_dep,'score']+ new_df.loc[filter_rest_dep,'score']*0.05

new_df=new_df.rename(columns={"score":"new_score"})
previous_score = df["score"]
new_score = new_df["new_score"]
score_increase = new_score-previous_score

# student_id, department, previous_score, new_score, score_increase
# add colums part
new_df["new_score"] = new_score
new_df["score_increase"]=score_increase
new_df["previous_score"] = previous_score

print(new_df[["student_id","department","previous_score","new_score","score_increase"]])
# print(df.info())

```

# Question 05: [ Marks 10 ]

Using Matplotlib, create a bar chart that shows the average score per department.

* X-axis: Department names
* Y-axis: Average score
* Add a title: "Average Score by Department"
* Display the chart

## Answer 05:

```python
import matplotlib.pyplot as plt


avg_dep_df = df.groupby("department")['score'].mean()

x_axix = avg_dep_df.index
y_axix = avg_dep_df.values

plt.bar(x_axix, y_axix)

plt.title("Average Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Score")


plt.xticks(rotation=90)
plt.show()
```
