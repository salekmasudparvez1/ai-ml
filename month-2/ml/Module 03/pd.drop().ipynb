# Pandas `drop()` Quick Notes

> Remove rows or columns from a DataFrame.

---

## 📦 Import

```python
import pandas as pd
```

---

## 🏗️ Create DataFrame

```python
df = pd.DataFrame({
    "name": ["A", "B", "C"],
    "age": [20, 21, 22],
    "salary": [50000, 60000, 70000]
})
```

---

## 📌 Drop Column

<Tabs>
<Tab title="Code">

```python
df.drop(columns=["salary"])
```

</Tab>

<Tab title="Output">

```python
  name  age
0    A   20
1    B   21
2    C   22
```

</Tab>
</Tabs>

---

## 📌 Drop Multiple Columns

<Tabs>
<Tab title="Code">

```python
df.drop(columns=["name", "salary"])
```

</Tab>

<Tab title="Output">

```python
   age
0   20
1   21
2   22
```

</Tab>
</Tabs>

---

## 📌 Drop Row

<Tabs>
<Tab title="Code">

```python
df.drop(index=[0])
```

</Tab>

<Tab title="Output">

```python
  name  age  salary
1    B   21   60000
2    C   22   70000
```

</Tab>
</Tabs>

---

## 📌 Drop Multiple Rows

```python
df.drop(index=[0, 2])
```

### Example

```python
df.drop(["salary"], axis=1)
```

---

## ⚡ inplace=True

Modify original DataFrame directly.

```python
df.drop(columns=["salary"], inplace=True)
```

---

## ⚡ Ignore Errors

```python
df.drop(columns=["abc"], errors="ignore")
```

---

## 🤖 AI/ML Example

```python
X = df.drop(columns=["target"])
y = df["target"]
```

---

## ✅ Important

- `drop()` → remove rows/columns
- `dropna()` → remove missing values
- Used in AI/ML preprocessing

---

## ⭐ Best Practice

### Good

```python
df.drop(columns=["A"])
```

### Old Style

```python
df.drop(["A"], axis=1)
```

---

## 🚀 Common Uses

- Remove ID columns
- Remove useless features
- Remove leakage columns
- Clean dataset
- Feature selection
