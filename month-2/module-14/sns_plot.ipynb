```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

# Lineplot using Seaborn

```python
student = pd.read_csv('sns_data.csv') 
student.head()
```

```python
sns.lineplot(x="week",y="attendance_rate", data=student,errorbar=None,hue="gender")
```

```python
#figure level

sns.relplot(kind ='line', data= student , x = 'week' , y = 'attendance_rate',errorbar=None,hue='gender')
```

```python
#figure level

sns.relplot(kind ='line', data= student , x = 'week' , y = 'attendance_rate',errorbar=None,hue='class_level')
```

```python
sns.scatterplot(data = student, x = 'study_hours' , y ='test_score',hue='gender',style='subject',size='Tshirt_size') 
```

```python

sns.relplot(kind='scatter', data = student , x = 'study_hours' , y ='test_score',col='gender',style='subject',size='Tshirt_size')
```

```python
student = pd.read_csv('student_dataset_complete.csv') 
student.head()
```

```python
# axes level
sns.histplot(data = student , x ='attendance_rate',hue='gender',bins=10,element='step') 
```

```python
#jointplot
sns.jointplot(data=student_marks , x ='study_hours' , y='Marks',hue='gender') 
```
